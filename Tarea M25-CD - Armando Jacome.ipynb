```python
#Ejercicio 1 | Obtener raiz cuadrada de un número
#Sin lambda
def raiz(num):
    return num**(0.5)

raiz(4)
```

```python
#Con lambda
raiz = lambda x: x**(0.5)
raiz(4)
```

```python
#Ejercicio 2 | Largo de una cadena de palabras
#Sin map
texto = "Hola, mundo, te tkm"
palabras = texto.split()

largo = []

for p in palabras:
    largo.append(len(p))

print(largo)
```

```python
#Con map
largo2 = list(map(len, palabras))
print(largo2)
```

```python
#Ejercicio 3 | Calcular el producto de una lista
#Sin Reduce
numeros = [1, 2, 3, 4, 5]

def producto_acum(numeros):
    resultado = 1
    for i in numeros:
        resultado *= i
    return resultado

print(producto_acum(numeros))
```

```python
#Con reduce
from functools import reduce
prod_acum = reduce((lambda i, n: i*n), numeros)
prod_acum
```

```python
#Ejercicio 4 | Encontrar palabras que contengan mayúsculas o números en un listado
#Sin filter
texto = "P. sherman, calle Wallaby 42, Sidney"
palabras = texto.split()

def filtro(lista):
    palabras_filt = []
    for p in lista:
        if any(c.isupper() or c.isdigit() for c in p):
            palabras_filt.append(p)
    return palabras_filt

print(filtro(palabras))
```

```python
#Con filter
filtro2 = list(filter(lambda x:any(i.isupper() or i.isdigit() for i in x),palabras))
filtro2
```
