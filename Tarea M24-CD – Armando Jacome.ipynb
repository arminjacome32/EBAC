```python
# Import data from archive
import pandas as pd
df = pd.read_csv("SYB66_246_202310_Population Growth, Fertility and Mortality Indicators.csv", encoding = "latin1")
df
```

```python
#Set column titles from the first row
df.columns = df.iloc[0]
df = df.iloc[1:].reset_index(drop=True)
df
```

```python
# Change second column name to "Name"
df.columns.values[1] = 'Name'
```

```python
df
```

```python
#Convert Value Column Data into float type
df["Value"] = df["Value"].str.replace(",", "", regex=False).astype(float)
df
```

```python
#Convert Year Column Data into Int
df["Year"].dtype
df["Year"] = df["Year"].astype(int)
```

```python
#Extract only 2020 data
df = df[df["Year"] == 2020]
df
```

```python
df.columns = df.columns.astype(str).str.strip()
```

```python
#Creating a Pivot Table from the resulting df: Region/Country/Area vs. Series, Data = "Value"
pTable = pd.pivot_table(
    data=df,
    values="Value",
    index=["Name"],
    columns=["Series"]
)
pTable
```

```python
#Creating graphs

import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots()
scatter = sns.regplot(data = pTable, 
                      x = "Infant mortality for both sexes (per 1,000 live births)", 
                      y = "Life expectancy at birth for both sexes (years)",
                      ax = ax
                     )

ax.set(title = "Infant Mortality Vs. Life expectancy at birth")

plt.show()
fig.savefig("Regresion_InfantMortalityVsLifeExpectancyAtBirth.png")
```

```python
from itertools import combinations
import textwrap

lista = ["Infant mortality for both sexes (per 1,000 live births)",
         "Life expectancy at birth for both sexes (years)",
        "Maternal mortality ratio (deaths per 100,000 population)",
        "Population annual rate of increase (percent)",
        "Total fertility rate (children per women)"
        ]

def scatters(a, b, ax):
    sns.regplot(x = a, y = b, data = pTable, ax = ax)
    titulo = f"{a} Vs. {b}"
    titulo = textwrap.fill(titulo, width = 40)
    ax.set_title(titulo)

fig, axes = plt.subplots(5,2, figsize = (18, 20))

for (a,b), ax in zip(combinations(lista, 2), axes.flat):
    scatters(a, b, ax)

plt.tight_layout()
plt.show()
fig.savefig("regresiones.png", dpi = 300)
```

```python
from itertools import combinations
import textwrap
import matplotlib.pyplot as plt
import seaborn as sns

lista = ["Infant mortality for both sexes (per 1,000 live births)",
        "Maternal mortality ratio (deaths per 100,000 population)",
        "Life expectancy at birth for both sexes (years)",
        "Population annual rate of increase (percent)",
        "Total fertility rate (children per women)"
        ]

def violin(a, ax):
    sns.violinplot(y = a, data = pTable, ax = ax)
    titulo = f"{a}"
    titulo = textwrap.fill(titulo, width = 40)
    ax.set_title(titulo)

plt.style.use("ggplot")
fig, axes = plt.subplots(3,2, figsize = (16, 18))

for a, ax in zip(lista, axes.flat):
    violin(a, ax)
    
plt.tight_layout()
plt.show()
fig.savefig("violines.png", dpi = 300)
```
