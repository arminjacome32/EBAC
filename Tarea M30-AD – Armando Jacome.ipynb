```python
#Todas las columnas numericas tienen formato de número, por lo que no es necesario.Pasaré la de Age y Potencial a float solo por práctica.

df[['Age', 'Potential']] = df[['Age', 'Potential']].astype(float)
df.dtypes
```

```python
#Nueva columna
df['Years Playing'] = 2026 - df['Joined']
df
```

```python
#Buscar jugadores de México:
df.loc[df['Nationality'] == 'Mexico']
```

```python
#Ordenar y mostrar los datos por la columna ReleaseClause (sueldo)

df.sort_values(by = 'Release Clause', ascending = False)
```

```python
#Generar un nuevo dataset que contenga el año (joined) y el número de jugadores (groupby)

df2 = df.groupby('Joined').count()['ID']
df2
```
