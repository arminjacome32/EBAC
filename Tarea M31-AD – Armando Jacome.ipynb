```python
#Generar la media, mínimo y máximo de los jugadores en total, y por país (groupby)
df[['Age', 'Value', 'Joined']].mean()
df[['Age', 'Value', 'Joined']].min()
df[['Age', 'Value', 'Joined']].max()
```

```python
#Generar la media, mínimo y máximo de los jugadores por país (groupby)
df.groupby('Nationality')[['Age', 'Value', 'Joined']].agg(['mean', 'min', 'max'])
```
