```python
import pandas as pd
import plotly.express as px

df = pd.read_csv('fifa_eda.csv')
df
```

```python
#Elemento 1: Obtener matriz de correlación y dejarla como heat map 
df1 = df[['Age', 'Overall', 'Potential', 'Value', 'Wage', 'International Reputation', 'Skill Moves', 'Joined', 'Height', 'Weight', 'Release Clause']]
df1
```

```python
corr = df1.corr()
corr
```

```python
px.imshow(corr, text_auto = True)
```

```python
#Elemento 2: Gráfico de Correlación que responda – “Cuál es la relación entre la edad y el overall?” – explicar la salida
df2 = df[['Age', 'Overall']]
df2
```

```python
df2.corr()
```

```python
fig2 = px.scatter(df2, x = 'Overall', y = 'Age', trendline = 'ols', title = 'Correlación Overall ')
fig2
```

```python
#La edad y el overall presentan una correlación lineal positiva y moderada (rho de spearman = .45)
```

```python
#Elemento 3: Generar un gráfico de barras por club que indique el número de jugadores 
df3 = df[['ID', 'Club']].groupby('Club').size().reset_index(name = 'Jugadores').sort_values(by = 'Jugadores', ascending = False)
df3
```

```python
fig3 = px.bar(df3, x = 'Club', y = 'Jugadores', title = 'Jugadores de FIFA por Club')
fig3
```

```python
#Elemento 4: Hacer un gráfico multipanel que indique la relación altura (height) vs skill moves, siendo la variable del panel si es zurdo o derecho
corr4 = df['Height'].corr(df['Skill Moves'])
corr4
```

```python
fig4 = px.scatter(df, x = 'Skill Moves', y = 'Height', trendline = 'ols', facet_col = 'Preferred Foot', title = 'Correlación Altura y Habilidades (vista por pie dominante)')
fig4
```

```python
#Elemento 5: Un gráfico que crea importante para mostrar a los potenciales cracks
fig5 = px.box(df, x = 'Potential', title = 'Potencial de los jugadores')
fig5


#El diagrama de caja permite identificar la distribución del potencial de los jugadores. Los valores atípicos superiores 
#representan observaciones cuyo potencial supera considerablemente al de la mayoría de los jugadores. 
#Por lo tanto, estos casos pueden ser de especial interés para identificar talentos prometedores (*cracks*) 
#que destacan por su potencial de desarrollo.
```
