```python
#Ejercicio 3 y 4 | Leer Archivo e Imprimir primeros 50 registros
with open('Height of Male and Female by Country 2022.csv', 'r',
         encoding = 'utf-8') as a:

    archivo = csv.reader(a)

    rows = []
    for row in archivo:
        rows.append(row)

rows[0:51]
```

```python
# Ejercicio 5 | Calcular Media, Minima y Máxima del Set Completo
from statistics import mean
import numpy as np
df = np.array(rows)
print(df)
```

```python
print(df.dtype)
```

```python
paises = df[1:, 1]
male_cm = df[1:, 2].astype(float)
male_ft = df[1:, 4].astype(float)
female_cm = df[1:, 3].astype(float)
female_ft = df[1:, 5].astype(float)
```

```python
print(f"Media hombres cm: {mean(male_cm):.2f}")
print(f"Media hombres ft: {mean(male_ft):.2f}")
print(f"Media mujeres cm: {mean(female_cm):.2f}")
print(f"Media mujeres ft: {mean(female_ft):.2f}")
```

```python
print(f"Min hombres cm: {np.min(male_cm):.2f}")
print(f"Min hombres ft: {np.min(male_ft):.2f}")
print(f"Min mujeres cm: {np.min(female_cm):.2f}")
print(f"Min mujeres ft: {np.min(female_ft):.2f}")
```

```python
print(f"Max hombres cm: {np.max(male_cm):.2f}")
print(f"Max hombres ft: {np.max(male_ft):.2f}")
print(f"Max mujeres cm: {np.max(female_cm):.2f}")
print(f"Max mujeres ft: {np.max(female_ft):.2f}")
```
