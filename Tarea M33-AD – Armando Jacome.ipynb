```python
df9 = df.copy()
df9['Year'] = pd.to_numeric(df9['Year'], errors = 'coerce')
px.bar(df9.groupby('Platform').size().sort_values(ascending = False),
      title = 'Videojuegos por Plataforma'
      )
```

```python
#Crecimiento en los últimos 4 años
# 1) Qué género ha tenido el mayor y menor crecimiento.
df10 = df9.copy()
df10 = df10.pivot_table(values = 'Global_Sales', index = 'Genre', columns = 'Year')
df10 = df10.fillna("")
df10[[2010,2015]]
df10['Growth'] = ((df10[2015] - df10[2011]) / df10[2011])*100
df10 = df10[[2011,2015,'Growth']].sort_values(by = 'Growth', ascending = False)
df10
```

```python
from plotly.subplots import make_subplots
import plotly.express as px
import plotly.graph_objects as go

fig = make_subplots(specs=[[{"secondary_y": True}]])

# Barras: ventas de 2011 y 2015
fig.add_trace(
    go.Bar(x=df10.index, y=df10[2011], name='2011'),
    secondary_y=False
)

fig.add_trace(
    go.Bar(x=df10.index, y=df10[2015], name='2015'),
    secondary_y=False
)

# Puntos: porcentaje de crecimiento
fig.add_trace(
    go.Scatter(
        x=df10.index,
        y=df10['Growth'],
        name='Crecimiento (%)',
        mode='lines+markers'
    ),
    secondary_y=True
)

fig.update_layout(barmode='overlay')

fig.update_yaxes(title_text='Ventas globales', secondary_y=False)
fig.update_yaxes(title_text='Crecimiento (%)', secondary_y=True)

fig.show()
```

```python
#Genero de mayor crecimiento por zona geográfica
df15 = df9.copy()
df15 = df15.melt(
    id_vars = 'Year',
    value_vars = ['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales'],
    var_name = 'Region',
    value_name = 'Sales'
)

df15 = df15.pivot_table(values = 'Sales', index = 'Region', columns = 'Year')
df16 = df15.copy()
df16['Growth'] = ((df16[2015] - df16[2011])/df16[2011])*100
df16 = df16[[2011, 2015, 'Growth']].sort_values(by = 'Growth', ascending = False)
df16
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

fig = make_subplots(specs=[[{"secondary_y": True}]])

# Barras de ventas
fig.add_trace(
    go.Bar(
        x=df16.index,
        y=df16[2015.0],
        name='2015'
    ),
    secondary_y=False
)

fig.add_trace(
    go.Bar(
        x=df16.index,
        y=df16[2011.0],
        name='2011'
    ),
    secondary_y=False
)

# Línea de crecimiento porcentual
fig.add_trace(
    go.Scatter(
        x=df16.index,
        y=df16['Growth'],
        name='Growth (%)',
        mode='lines+markers'
    ),
    secondary_y=True
)

# Configuración
fig.update_layout(
    title='Ventas y crecimiento por región',
    barmode='overlay',
    xaxis_title='Región'
)

fig.update_yaxes(title_text='Ventas globales', secondary_y=False)
fig.update_yaxes(title_text='Crecimiento (%)', secondary_y=True)

fig.show()
```
